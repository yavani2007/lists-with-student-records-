```python
# Student names and marks

names = ["yavanika", "janaki", "Ram", "Priya"]
marks = [85, 92, 78, 88]

# Display records
print("Student Records:")
for i in range(len(names)):
    print(names[i], "-", marks[i])

# Search student
search_name = input("\nEnter student name to search: ")

if search_name in names:
    index = names.index(search_name)
    print(search_name, "scored", marks[index])
else:
    print("Student not found")

# Sort marks
sorted_marks = sorted(marks)
print("\nMarks in ascending order:", sorted_marks)

# Highest and lowest score
highest = max(marks)
lowest = min(marks)

print("Highest Score:", highest)
print("Lowest Score:", lowest)

# Student with highest score
high_index = marks.index(highest)
print("Top Student:", names[high_index])
```
